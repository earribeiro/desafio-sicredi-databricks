```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

CATALOG = "workspace"
SCHEMA = "desafio_sicredi"

BASE = f"{CATALOG}.{SCHEMA}"
```

```python
##ler_os_dados

df_transacoes = spark.table(
    f"{BASE}.bronze_transacoes_cartao"
)

df_estornos = spark.table(
    f"{BASE}.bronze_estornos"
)

df_cartoes_validos = spark.table(
    f"{BASE}.silver_cartoes"
)

print("Transações:", df_transacoes.count())
print("Estornos:", df_estornos.count())
print("Cartões válidos:", df_cartoes_validos.count())
```

```python
##conversao_dados

df_transacoes_tipadas = (
    df_transacoes
    .withColumn(
        "id_transacao",
        F.expr("try_cast(id_transacao AS BIGINT)")
    )
    .withColumn(
        "id_cartao",
        F.expr("try_cast(id_cartao AS BIGINT)")
    )
    .withColumn(
        "data_transacao",
        F.expr("try_cast(data_transacao AS TIMESTAMP)")
    )
    .withColumn(
        "valor",
        F.expr("try_cast(valor AS DECIMAL(18,2))")
    )
)

display(df_transacoes_tipadas)
```

```python
##deduplicar_dados

janela_transacoes = (
    Window
    .partitionBy("id_transacao")
    .orderBy(
        F.col("timestamp_ingestao").desc_nulls_last(),
        F.col("hash_linha").desc()
    )
)

df_transacoes_dedup = (
    df_transacoes_tipadas
    .withColumn(
        "rn",
        F.row_number().over(janela_transacoes)
    )
    .filter(F.col("rn") == 1)
    .drop("rn")
)

print(
    "Transações após deduplicação:",
    df_transacoes_dedup.count()
)
```

```python
##validar_integridade

df_cartoes_ids = (
    df_cartoes_validos
    .select("id_cartao")
    .distinct()
    .withColumn(
        "cartao_existe",
        F.lit(True)
    )
)

df_transacoes_integridade = (
    df_transacoes_dedup
    .join(
        df_cartoes_ids,
        on="id_cartao",
        how="left"
    )
)

display(
    df_transacoes_integridade.select(
        "id_transacao",
        "id_cartao",
        "valor",
        "cartao_existe"
    )
)
```

```python
df_transacoes_qualidade = (
    df_transacoes_integridade
    .withColumn(
        "motivo_erro",
        F.when(
            F.col("id_transacao").isNull(),
            "ID_TRANSACAO_INVALIDO"
        )
        .when(
            F.col("id_cartao").isNull(),
            "ID_CARTAO_INVALIDO"
        )
        .when(
            F.col("cartao_existe").isNull(),
            "CARTAO_INEXISTENTE_OU_INVALIDO"
        )
        .when(
            F.col("data_transacao").isNull(),
            "DATA_TRANSACAO_INVALIDA"
        )
        .when(
            F.col("valor").isNull() |
            (F.col("valor") <= 0),
            "VALOR_INVALIDO"
        )
        .when(
            F.col("nome_estabelecimento").isNull() |
            (F.trim(F.col("nome_estabelecimento")) == ""),
            "ESTABELECIMENTO_OBRIGATORIO"
        )
    )
)
```

```python
##separar_transacaoes_validas

df_transacoes_validas = (
    df_transacoes_qualidade
    .filter(F.col("motivo_erro").isNull())
    .drop("motivo_erro", "cartao_existe")
)

df_transacoes_quarentena = (
    df_transacoes_qualidade
    .filter(F.col("motivo_erro").isNotNull())
    .drop("cartao_existe")
)

print(
    "Transações válidas:",
    df_transacoes_validas.count()
)

print(
    "Transações em quarentena:",
    df_transacoes_quarentena.count()
)
```

```python
df_estornos_tipados = (
    df_estornos
    .withColumn(
        "id_estorno",
        F.expr("try_cast(id_estorno AS BIGINT)")
    )
    .withColumn(
        "id_transacao",
        F.expr("try_cast(id_transacao AS BIGINT)")
    )
    .withColumn(
        "data_estorno",
        F.expr("try_cast(data_estorno AS TIMESTAMP)")
    )
)
```

```python
df_transacoes_ids = (
    df_transacoes_validas
    .select("id_transacao")
    .distinct()
    .withColumn(
        "transacao_existe",
        F.lit(True)
    )
)

df_estornos_integridade = (
    df_estornos_tipados
    .join(
        df_transacoes_ids,
        on="id_transacao",
        how="left"
    )
)

df_estornos_qualidade = (
    df_estornos_integridade
    .withColumn(
        "motivo_erro",
        F.when(
            F.col("id_estorno").isNull(),
            "ID_ESTORNO_INVALIDO"
        )
        .when(
            F.col("id_transacao").isNull(),
            "ID_TRANSACAO_INVALIDO"
        )
        .when(
            F.col("data_estorno").isNull(),
            "DATA_ESTORNO_INVALIDA"
        )
        .when(
            F.col("transacao_existe").isNull(),
            "TRANSACAO_INEXISTENTE_OU_INVALIDA"
        )
    )
)
```

```python
df_estornos_validos = (
    df_estornos_qualidade
    .filter(F.col("motivo_erro").isNull())
    .drop("motivo_erro", "transacao_existe")
)

df_estornos_quarentena = (
    df_estornos_qualidade
    .filter(F.col("motivo_erro").isNotNull())
    .drop("transacao_existe")
)

print("Estornos válidos:", df_estornos_validos.count())
print("Estornos em quarentena:", df_estornos_quarentena.count())
```

```python
##transacao_estornada

df_estornos_ids = (
    df_estornos_validos
    .select("id_transacao")
    .distinct()
    .withColumn("estornada", F.lit(True))
)

df_transacoes_final = (
    df_transacoes_validas
    .join(
        df_estornos_ids,
        on="id_transacao",
        how="left"
    )
    .withColumn(
        "estornada",
        F.coalesce(
            F.col("estornada"),
            F.lit(False)
        )
    )
    .withColumn(
        "valor_liquido",
        F.when(
            F.col("estornada"),
            F.lit(0).cast("decimal(18,2)")
        ).otherwise(F.col("valor"))
    )
)

display(
    df_transacoes_final.select(
        "id_transacao",
        "valor",
        "estornada",
        "valor_liquido"
    ).orderBy("id_transacao")
)
```

```python
tabelas = {
    "silver_transacoes_cartao": df_transacoes_final,
    "quarentena_transacoes_cartao": df_transacoes_quarentena,
    "silver_estornos": df_estornos_validos,
    "quarentena_estornos": df_estornos_quarentena
}

for nome, df in tabelas.items():
    (
        df.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(f"{BASE}.{nome}")
    )

    print(f"Tabela gravada: {BASE}.{nome}")
```

```python
%sql

SELECT
    COUNT(*) AS total_transacoes_validas,
    SUM(valor) AS valor_bruto,
    SUM(valor_liquido) AS valor_liquido,
    SUM(CASE WHEN estornada THEN 1 ELSE 0 END)
        AS total_estornadas
FROM workspace.desafio_sicredi.silver_transacoes_cartao;
```
