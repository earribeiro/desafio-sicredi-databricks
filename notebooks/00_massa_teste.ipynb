```python
base_path = (
    "/Volumes/workspace/desafio_sicredi/"
    "arquivos_entrada"
)

arquivos = {
    "clientes": df_clientes,
    "contas": df_contas,
    "cartoes": df_cartoes,
    "estornos": df_estornos
}

for nome, df in arquivos.items():
    (
        df.coalesce(1)
        .write
        .mode("overwrite")
        .option("header", True)
        .csv(f"{base_path}/{nome}")
    )

print("Arquivos cadastrais gravados!")
```

```python
datas = ["20261013", "20261014", "20261015"]

for data in datas:
    df_dia = (
        df_transacoes
        .filter(F.col("data_referencia") == data)
        .drop("data_referencia")
    )

    (
        df_dia.coalesce(1)
        .write
        .mode("overwrite")
        .option("header", True)
        .csv(
            f"{base_path}/"
            f"transacoes_cartao_{data}"
        )
    )

print("Transações dos três dias gravadas!")
```
